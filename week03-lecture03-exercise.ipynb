```python

# Task 1: CO2 per capita — highlight the United States

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Load dataset
df = pd.read_csv("/kaggle/input/datasets/josephsudheer0506/co2-emission/co2_emissions.csv")

highlight_country = "United States"

# Create figure
fig = go.Figure()

# Plot all other countries in light grey
for country in df["Country"].unique():
    country_df = df[df["Country"] == country].sort_values("Year")

    if country != highlight_country:
        fig.add_trace(go.Scatter(
            x=country_df["Year"],
            y=country_df["CO2_per_capita"],
            mode="lines",
            line=dict(color="#EEEEEE", width=1.5),
            hovertemplate=(
                country + "<br>Year: %{x}"
                + "<br>CO2 per capita: %{y:.2f}<extra></extra>"
            ),
            showlegend=False
        ))

# Highlight the United States
us_df = df[df["Country"] == highlight_country].sort_values("Year")

fig.add_trace(go.Scatter(
    x=us_df["Year"],
    y=us_df["CO2_per_capita"],
    mode="lines",
    line=dict(color="#2563EB", width=3),
    name=highlight_country,
    showlegend=False,
    hovertemplate=(
        "United States<br>Year: %{x}"
        + "<br>CO2 per capita: %{y:.2f}<extra></extra>"
    )
))

# Add direct label at the end of the highlighted line
last_row = us_df.iloc[-1]

fig.add_annotation(
    x=last_row["Year"],
    y=last_row["CO2_per_capita"],
    text="United States",
    showarrow=False,
    xshift=8,
    xanchor="left",
    font=dict(color="#2563EB", size=12)
)

# Styling
fig.update_layout(
    title=dict(
        text="<b>US CO₂ per capita emissions remain high</b>",
        x=0.02,
        xanchor="left"
    ),
    xaxis_title="Year",
    yaxis_title="CO₂ emissions per capita",
    template="plotly_white",
    plot_bgcolor="white",
    paper_bgcolor="white",
    margin=dict(l=60, r=130, t=80, b=50)
)

fig.update_xaxes(
    showline=False,
    showgrid=False,
    zeroline=False,
    dtick=5
)

fig.update_yaxes(
    showline=False,
    showgrid=True,
    gridcolor="#F0F0F0",
    zeroline=False
)

fig.show()
```

```python

# Task 2: Asian countries — highlight China

highlight_country = "China"

# Filter Asian countries
asia_df = df[df["Region"] == "Asia"].copy()

fig = go.Figure()

# Plot all other Asian countries in grey
for country in asia_df["Country"].unique():
    country_df = (
        asia_df[asia_df["Country"] == country]
        .sort_values("Year")
    )

    if country != highlight_country:
        fig.add_trace(go.Scatter(
            x=country_df["Year"],
            y=country_df["CO2_Mt"],
            mode="lines",
            line=dict(color="#EEEEEE", width=1.5),
            hovertemplate=(
                country + "<br>Year: %{x}"
                + "<br>CO₂: %{y:,.1f} Mt<extra></extra>"
            ),
            showlegend=False
        ))

# Highlight China
china_df = (
    asia_df[asia_df["Country"] == highlight_country]
    .sort_values("Year")
)

fig.add_trace(go.Scatter(
    x=china_df["Year"],
    y=china_df["CO2_Mt"],
    mode="lines",
    line=dict(color="#DC2626", width=3),
    showlegend=False,
    hovertemplate=(
        "China<br>Year: %{x}"
        + "<br>CO₂: %{y:,.1f} Mt<extra></extra>"
    )
))

# Label China at the end of its line
last_row = china_df.iloc[-1]

fig.add_annotation(
    x=last_row["Year"],
    y=last_row["CO2_Mt"],
    text="China",
    showarrow=False,
    xshift=8,
    xanchor="left",
    font=dict(color="#DC2626", size=12)
)

# Styling
fig.update_layout(
    title=dict(
        text="<b>China dominates Asia's total CO₂ emissions</b>",
        x=0.02,
        xanchor="left"
    ),
    xaxis_title="Year",
    yaxis_title="CO₂ emissions (million tonnes)",
    template="plotly_white",
    plot_bgcolor="white",
    paper_bgcolor="white",
    margin=dict(l=60, r=100, t=80, b=50)
)

fig.update_xaxes(
    showline=False,
    showgrid=False,
    zeroline=False,
    dtick=5
)

fig.update_yaxes(
    showline=False,
    showgrid=True,
    gridcolor="#F0F0F0",
    zeroline=False
)

fig.show()
```
